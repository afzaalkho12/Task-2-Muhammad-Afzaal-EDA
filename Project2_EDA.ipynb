# Data Cleaning

In this section, we will clean the dataset by handling missing values, verifying data types, checking data consistency, and preparing the data for analysis.

```python
# Data Types

df.dtypes
```

```python
# Convert Date Column

df["Date"] = pd.to_datetime(df["Date"])

print("Date column converted successfully!")
```

```python
# Handle Missing Coupon Codes

df["CouponCode"] = df["CouponCode"].fillna("No Coupon")

print("Missing Coupon Codes handled successfully!")
```

```python
# Verify Missing Values

df.isnull().sum()
```

```python
# Feature Engineering

df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month_name()
df["MonthNumber"] = df["Date"].dt.month
df["Quarter"] = df["Date"].dt.quarter
df["Day"] = df["Date"].dt.day_name()

print("New features created successfully!")
```

```python
df.head()
```

## Analysis 1: Monthly Orders

This analysis examines the number of customer orders placed each month. It helps identify seasonal trends, peak business periods, and fluctuations in customer purchasing behavior.

### Objective
- Determine the total number of orders placed each month.
- Identify months with the highest and lowest order volumes.
- Understand monthly ordering trends.

### Visualization
A bar chart is used to compare the number of orders across different months.

---

### Interpretation

The bar chart displays the monthly distribution of customer orders. Higher bars indicate months with increased customer activity, while lower bars indicate periods with fewer orders.

### Key Insight

This analysis helps identify seasonal demand patterns and supports better inventory management, staffing, and marketing strategies.

```python
# Monthly Orders

monthly_orders = df.groupby("Month")["OrderID"].count()

monthly_orders = monthly_orders.reindex([
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
])

print(monthly_orders)
```

```python
plt.figure(figsize=(10,5))

monthly_orders.plot(kind="bar")

plt.title("Number of Orders by Month")
plt.xlabel("Month")
plt.ylabel("Orders")
plt.xticks(rotation=45)

plt.savefig("../Charts/monthly_orders.png", dpi=300, bbox_inches="tight")

plt.show()
```

## Analysis 3: Top 10 Best Selling Products

## Analysis 11: Correlation Analysis

This analysis measures the relationship between numerical variables in the dataset.

### Objective
- Identify relationships among numerical variables.
- Detect positive and negative correlations.
- Support data-driven decision making.

### Visualization
A correlation heatmap is used to visualize relationships between numerical features.

---

### Interpretation

The heatmap displays correlation coefficients ranging from -1 to +1. Values closer to +1 indicate strong positive relationships, while values closer to -1 indicate strong negative relationships.

### Key Insight
